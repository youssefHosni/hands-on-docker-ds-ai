# House prices

A first look at `data/houses.csv`, a small made-up dataset.

```python
import pandas as pd

houses = pd.read_csv("../data/houses.csv")
houses.head()
```

```python
houses.plot.scatter(x="size_m2", y="price_k", title="Price vs size");
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression().fit(houses[["size_m2", "rooms", "age_years"]], houses["price_k"])
print(f"Price per extra m2: {model.coef_[0]:.2f}k")
```
